```python
import json

json_path = "/content/5WQA_QA_Pair.json"  # Update the path if needed

# Load JSON data
with open(json_path, "r", encoding="utf-8") as f:
    qa_pairs = json.load(f)

# Print dataset details
print(f"✅ Successfully loaded JSON with {len(qa_pairs)} entries.")
print("🔹 Sample QA Pair:\n", json.dumps(qa_pairs[:2], indent=2))
```

```python
# Remove 'qa_evidence_t53b' from each entry
for entry in qa_pairs:
    entry.pop("qa_evidence_t53b", None)  # 'None' prevents errors if the key is missing

print(f"✅ Column 'qa_evidence_t53b' removed. Total entries: {len(qa_pairs)}")
print("🔹 Sample after removal:\n", json.dumps(qa_pairs[:2], indent=2))
```

```python
import random

# Select 100 random entries
qa_pairs_sample = random.sample(qa_pairs, 100)

print(f"✅ Extracted {len(qa_pairs_sample)} random entries.")
print("🔹 Sample data:\n", json.dumps(qa_pairs_sample[:3], indent=2))
```

```python
# Define function to label questions
def label_qa_pair(question):
    question = question.lower()
    if question.startswith("what"):
        return "What"
    elif question.startswith("when"):
        return "When"
    elif question.startswith("why"):
        return "Why"
    elif question.startswith("where"):
        return "Where"
    elif question.startswith("who"):
        return "Who"
    else:
        return "Other"  # In case some don't fit the 5Ws

# Add labels to each QA pair
for entry in qa_pairs_sample:
    entry["label"] = label_qa_pair(entry["qq_pophetnet"])

print(f"✅ Labels assigned to {len(qa_pairs_sample)} entries.")
print("🔹 Sample labeled data:\n", json.dumps(qa_pairs_sample[:5], indent=2))
```

```python
# Count null values
null_questions = sum(1 for entry in qa_pairs_sample if not entry.get("qq_pophetnet"))
null_answers = sum(1 for entry in qa_pairs_sample if not entry.get("qa_srl"))

print(f"🔍 Null values check:")
print(f"❌ Missing 'qq_pophetnet' (questions): {null_questions}")
print(f"❌ Missing 'qa_srl' (answers): {null_answers}")
```

```python
# Count entries labeled as "Other"
other_count = sum(1 for entry in qa_pairs_sample if entry["label"] == "Other")

print(f"🔍 Total 'Other' labels found: {other_count} out of {len(qa_pairs_sample)}")
```

```python
# Get samples labeled as "Other"
other_samples = [entry for entry in qa_pairs_sample if entry["label"] == "Other"]

# Print a few examples
print(f"🔍 Found {len(other_samples)} entries labeled as 'Other'.")
print("🔹 Sample entries:\n", json.dumps(other_samples[:5], indent=2))
```

```python
import json
import random

# Load full dataset from JSON file
with open("/content/5WQA_QA_Pair.json", "r") as file:
    full_qa_dataset = json.load(file)

# Function to label QA pairs based on keywords
def assign_5w_label(question):
    if not question:  # If None or empty string
        return "Other"

    question_lower = question.lower()
    if "what" in question_lower:
        return "What"
    elif "when" in question_lower:
        return "When"
    elif "why" in question_lower:
        return "Why"
    elif "where" in question_lower:
        return "Where"
    elif "who" in question_lower:
        return "Who"
    else:
        return "Other"

# Assign labels to all QA pairs
for entry in full_qa_dataset:
    entry["label"] = assign_5w_label(entry.get("qq_pophetnet"))  # Use .get() to avoid KeyError

# Filter only entries with valid 5W labels
valid_qa_pairs = [entry for entry in full_qa_dataset if entry["label"] in ["What", "When", "Why", "Where", "Who"]]

# Randomly select 100 entries
qa_pairs_final = random.sample(valid_qa_pairs, 100)

# Print confirmation
print(f"✅ Selected {len(qa_pairs_final)} valid QA pairs (excluding 'Other').")
print("🔹 Sample labeled QA pairs:\n", json.dumps(qa_pairs_final[:5], indent=2))

# Save the final 100 labeled samples
output_path = "/content/5WQA_QA_Pair_100.json"
with open(output_path, "w") as output_file:
    json.dump(qa_pairs_final, output_file, indent=2)

print(f"✅ Final 100 labeled QA pairs saved as '{output_path}'.")
```

```python
from collections import Counter

# Check for null values
null_qq = sum(1 for entry in qa_pairs_final if entry.get("qq_pophetnet") is None)
null_label = sum(1 for entry in qa_pairs_final if entry.get("label") is None)

# Count occurrences of each label
label_counts = Counter(entry["label"] for entry in qa_pairs_final)

# Print results
print(f"✅ Null values in 'qq_pophetnet': {null_qq}")
print(f"✅ Null values in 'label': {null_label}")
print(f"\n🔹 Label Distribution:")
for label, count in label_counts.items():
    print(f"  {label}: {count}")

```
